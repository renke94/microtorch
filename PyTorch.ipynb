```python
from torch import nn
```

```python
class Model(nn.Module):
    def __init__(self):
        super().__init__()
        self.mlp = nn.Sequential(
            nn.Linear(64, 128),
            nn.Linear(128, 256),
        )

    def forward(self, x):
        return self.mlp(x)
```

```python
model = Model()
```
